## 2. Intended Use and Limits

### Intended Use

This playbook is intended to provide decision-support for prioritizing
pages that may deserve content review or refresh.

The ranking uses observed historical search-performance and content
signals to identify pages that appear more suitable for review than
others.

The output is a prioritization aid, not an automatic content decision.

### How the output should be used

1. Review the highest-ranked pages first.
2. Check the reason code for each recommendation.
3. Inspect the page and its current content manually.
4. Confirm that the recommendation makes sense in context.
5. Decide whether a content refresh is actually appropriate.
6. Record the human decision before making any content change.

### Limits

The score is based on historical data and available signals. It does
not prove that refreshing a page will increase traffic, rankings, CTR,
or conversions.

The ranking should be treated as directional decision-support.

The model may miss important information that is not represented in the
available dataset, such as search intent, content quality, business
priorities, seasonality, competitors, technical SEO issues, or recent
changes to a page.

The output should therefore not be interpreted as a guaranteed
prediction of future performance.

### What this playbook does NOT do

- It does not automatically rewrite content.
- It does not automatically publish changes.
- It does not delete or redirect pages.
- It does not guarantee traffic or ranking improvements.
- It does not establish causal effects.
- It does not replace human SEO/content judgment.

### Appropriate claim language

Use language such as:

- "observed"
- "measured"
- "directional"
- "may warrant review"
- "decision-support"
- "historical signals suggest"

Avoid claims such as:

- "will increase traffic"
- "will improve rankings"
- "guarantees better CTR"
- "automatically identifies the best pages"

## 3. Human Review + No-Go List

### Human Review Rules

Every high-priority recommendation must be reviewed by a person before
any content change is made.

The reviewer should:

1. Open the recommended page.
2. Read the reason code and understand why it was ranked.
3. Check the current content and search intent.
4. Consider business importance and page purpose.
5. Check for recent changes or unusual circumstances.
6. Decide whether a refresh is appropriate.
7. Record the final decision.

The model score is a prioritization signal, not a final decision.

### Cases Requiring Extra Review

Human review should be especially careful when:

- A page has very high traffic or business importance.
- Search data appears unusual or incomplete.
- The page has recently been changed.
- The recommendation conflicts with subject-matter knowledge.
- The page targets an important commercial or strategic topic.
- Multiple signals disagree.
- The page appears to be affected by seasonality or an external event.

### No-Go List

The system must NOT automatically:

- Publish rewritten content.
- Delete pages.
- Redirect URLs.
- Change important SEO metadata.
- Change canonical URLs.
- Modify robots.txt.
- Change internal linking across the website.
- Make claims about guaranteed traffic or ranking improvements.
- Treat the score as proof of causation.
- Override a human content or SEO decision.
- Take irreversible actions based only on the model score.

### Human-in-the-Loop Principle

The playbook is designed to answer:

"Which pages should I look at first?"

It is not designed to answer:

"Which changes should the system make automatically?"

The final content decision remains with a human reviewer.

## 4. Monitoring / Retrain Triggers

This playbook is intended as a lightweight decision-support system, so
monitoring should focus on whether the signals and recommendations
remain useful rather than assuming the model will remain correct
forever.

### What to Monitor

#### 1. Data availability
Check whether the required search-performance fields are still being
collected and whether missing values increase.

#### 2. Signal distribution
Monitor major inputs such as impressions, clicks, CTR, and average
position for large changes in their distributions.

#### 3. Recommendation distribution
Track how many pages receive each reason code and action.

A sudden change in the proportion of:
- CTR_OPPORTUNITY
- HIGH_DEMAND
- RANKING_OPPORTUNITY
- LOWER_PRIORITY

may indicate a change in the underlying data or website behavior.

#### 4. Human review outcomes
Track whether reviewers agree that high-ranked pages were reasonable
candidates for review.

The review outcome is useful feedback about whether the ranking remains
practically useful.

### Retrain / Recalibration Triggers

The model should be reviewed or retrained when:

- The underlying data definition changes.
- Important input signals are removed or changed.
- Search behavior changes substantially.
- The distribution of key features changes significantly.
- Recommendation patterns become unusually concentrated.
- Human reviewers repeatedly reject high-ranked recommendations.
- Model performance falls materially compared with the established
  baseline on a valid evaluation set.
- A significant amount of new historical data becomes available.

### Monitoring Frequency

For this non-production prototype, monitoring can be performed
periodically rather than continuously.

A reasonable approach is to review the signals and recommendation
distribution when a new evaluation dataset is available and before
reusing the playbook for a substantially different time period or
client population.

### Important Limitation

These triggers are monitoring guidelines, not automatic production
alerts. The current system is a research and decision-support
prototype. A human should decide whether the model needs
retraining, recalibration, or redesign.
